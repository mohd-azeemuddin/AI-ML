```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Azeem", "Rahul", "Priya", "John"],
    "Age": [23, 22, 24, 26],
    "Marks": [92, 81, 88, 75]
})
df[df["Marks"]>80]
```

```python
df[(df["Marks"] > 80) & (df["Age"] < 24)]
```

```python
df[(df["Marks"] > 90) | (df["Age"] < 23)]
```

```python
df["Passed"] = df["Marks"] > 80
```

```python
df["Grade"] = df["Marks"].apply(lambda x: "A" if x > 85 else "B")
print(df)
```

```python
df.sort_values(by="Marks")
```

```python
df.sort_values(by="Marks", ascending=False)
```
