```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Hours":[1,2,3,4,5,6],
    "Marks":[35,45,55,65,80,95]
}
df = pd.DataFrame(data)
print(df)
plt.figure(figsize=(6,4))
plt.plot(df["Hours"], df["Marks"], marker='o')
plt.title("Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["Hours"], df["Marks"])
plt.title("Hours vs Marks (Scatter Plot)")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
subjects = ["Math", "Science", "English", "AI"]
marks = [90, 85, 95, 98]

plt.figure(figsize=(6,4))
plt.bar(subjects, marks)
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Employee Dataset
data = {
    "Age": [22, 25, 28, 30, 35, 40, 45, 50],
    "Salary": [25000, 30000, 40000, 50000, 65000, 80000, 95000, 200000],
    "Experience": [0, 2, 4, 6, 9, 12, 15, 20],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "Management"]
}

df = pd.DataFrame(data)

# 1. Display Dataset
print("Employee Dataset:")
print(df)


# 2. Salary Distribution (Histogram)
plt.figure(figsize=(6,4))
plt.hist(df["Salary"], bins=5, edgecolor='black')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()

# 3. Experience vs Salary (Scatter)
plt.figure(figsize=(6,4))
plt.scatter(df["Experience"], df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.grid(True)
plt.show()

# 4. Compare Department Salaries
dept_salary = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(6,4))
plt.bar(dept_salary.index, dept_salary.values)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.grid(axis='y')
plt.show()

# 5. Find Possible Outliers
plt.figure(figsize=(6,4))
plt.boxplot(df["Salary"])
plt.title("Salary Outliers")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```
